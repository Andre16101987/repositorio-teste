```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('funcionarios_dados_ti.csv', sep = ';')
df = df[['Nome','Cargo','Salario']]
df.head(5)
```

```python
array_salario = np.array(df['Salario'])
array_salario
```

```python
media_salario = np.mean(array_salario)
media_salario
```

```python
mediana_salario = np.median(array_salario)
mediana_salario
```

```python
distancia_salario = (media_salario - mediana_salario)/(mediana_salario)
distancia_salario
```
