```python
import pandas as pd
```

```python
df2_2024 = pd.read_csv('2_2024.csv', sep=';')
df1_2024 = pd.read_csv('ca-2024-01.csv', sep=';')
df2_2023 = pd.read_csv('ca-2023-02.csv', sep=';')
df1_2023 = pd.read_csv('ca-2023-01.csv', sep=';')
df2_2022 = pd.read_csv('ca-2022-02.csv', sep=';')
df1_2022 = pd.read_csv('ca-2022-01.csv', sep=';')
df2_2021 = pd.read_csv('ca-2021-02.csv', sep=';', encoding= 'latin1')
df1_2021 = pd.read_csv('ca-2021-01.csv', sep=';')
df2_2020 = pd.read_csv('ca-2020-02.csv', sep=';')
df1_2020 = pd.read_csv('ca-2020-01.csv', sep=';')
```

```python
df = pd.concat([df2_2024,df1_2024,df2_2023,df1_2023,df2_2022,df1_2022,df2_2021, df1_2021, df2_2020, df1_2020])

```

```python
df['Valor de Venda'] = df['Valor de Venda'].str.replace(',','.').astype(float)
```

```python
df['Data da Coleta'] = pd.to_datetime(df['Data da Coleta'], format= "%d/%m/%Y")

df['mes'] = df['Data da Coleta'].dt.month
df['ano'] = df['Data da Coleta'].dt.year
```

```python
df_gasolina = df.loc[df['Produto'] == 'GASOLINA']
df_gasolina
```

```python
df_gasolina_media_regiao = df_gasolina.groupby(['mes','Estado - Sigla'])['Valor de Venda'].mean().reset_index()
df_gasolina_media_regiao
```

```python
df_gasolina_adt = df.loc[df['Produto'] == 'GASOLINA ADITIVADA']
df_gasolina_adt
```

```python
df_gasolina_adt_media = df_gasolina_adt.groupby(['Produto'])['Valor de Venda'].mean().reset_index()
df_gasolina_adt_media
```

```python
df_gasolina_media = df_gasolina.groupby(['Produto'])['Valor de Venda'].mean().reset_index()
df_gasolina_media
```

```python
df_gasosa = pd.concat([df_gasolina_media,df_gasolina_adt_media])
df_gasosa
```

```python
df_produto_media_ano = df_gasolina.groupby(['ano','Regiao - Sigla'])['Valor de Venda'].mean().reset_index()
df_produto_media_ano
```
