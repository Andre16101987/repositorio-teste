```python
import pandas as pd
import numpy as np
```

```python
df_base = pd.read_csv('https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv', sep = ';', encoding = 'latin1')
df_base
```

```python
metropolitana = ['Belford Roxo', 'Cachoeiras de Macacu', 'Duque de Caxias', 'Guapimirim', 'Itaboraí', 'Itaguaí', 'Japeri', 'Magé', 'Maricá', 'Mesquita', 'Nilópolis', 'Niterói', 'Nova Iguaçu', 'Paracambi', 'Petrópolis', 'Queimados', 'Rio Bonito', 'Rio de Janeiro', 'São Gonçalo', 'São João de Meriti','Tanguá']
```

```python
df_agora = df_base.loc[(df_base['ano'] == 2024) & (df_base['munic'].isin(metropolitana))]
df_agora
```

```python
df_rj= df_agora.groupby('munic')['furto_celular'].sum().reset_index()
df_rj
```

```python
array_rj = np.array(df_rj['furto_celular'])
array_rj
```

```python
media= np.mean(array_rj)
media
```

```python
mediana= np.median(array_rj)
mediana
```

```python
distancia= (media - mediana)/(mediana)
distancia
```
