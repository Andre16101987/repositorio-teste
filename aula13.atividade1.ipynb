```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
atividade_pesquisa = pd.read_csv('pesquisa_satisfacao.csv', sep= ';')
atividade_pesquisa
```

```python
clientes = np.array(atividade_pesquisa['Tempo de Espera (min)'])
clientes
```

```python
q1 = np.percentile(clientes,25)
print(f'Os mais rápidos são (q1): {q1}')
```

```python
primeiro_quartil = atividade_pesquisa[atividade_pesquisa['Tempo de Espera (min)'] <= q1]
primeiro_quartil = primeiro_quartil.sort_values(by= 'Tempo de Espera (min)')
primeiro_quartil
```

```python
primeiro_quartil.plot(kind = 'bar', x='Cliente',y='Tempo de Espera (min)')
plt.xlabel('Clientes')
plt.ylabel('Tempo de Espera')
plt.title('Cliente x Tempo de espera (min)')
plt.show()
```
