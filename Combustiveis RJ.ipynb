```python
import pandas as pd 

```

```python
df = pd.read_csv('2º_2024.csv', sep= ';')
df['Valor de Venda'] = df['Valor de Venda'].str.replace(',','.').astype(float)
```

```python
df_RJ = df.loc[(df['Estado - Sigla'] == 'RJ') & (df['Municipio'] == 'RIO DE JANEIRO')] # Acesso por condição
df_RJ 
```

```python
df_etanol = df_RJ.loc[df['Produto'] == 'ETANOL']

df_etanol_media_bairro = df_etanol.groupby('Bairro')['Valor de Venda'].mean().reset_index()
df_etanol_media_bairro = df_etanol_media_bairro.sort_values(by='Valor de Venda')

df_etanol_media_bairro.head(5)
```

```python
df_etanol_media_bairro.tail(5)
```

```python
df_gasolina = df_RJ.loc[df['Produto'] == 'GASOLINA']
df_gasolina_media_bairro = df_gasolina.groupby('Bairro')['Valor de Venda'].mean().reset_index()
df_gasolina_media_bairro = df_gasolina_media_bairro.sort_values(by='Valor de Venda')

df_gasolina_media_bairro.head(5)
```

```python
df_gasolina_media_bairro.tail(5)
```
