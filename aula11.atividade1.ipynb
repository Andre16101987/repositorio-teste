```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_caramelo = pd.read_csv('CARAMELO.csv',sep = ';')
```

```python
df_caramelo['Valor de Venda'] = df_caramelo['Valor de Venda'].str.replace(',','.').astype(float)
```

```python
df_estado_rj = df_caramelo.loc[df_caramelo['Estado Loja'] == 'RJ']
df_estado_rj
```

```python
df_cidade_loja = df_estado_rj.groupby(['Cidade Loja'])['Valor de Venda'].sum().reset_index()
df_cidade_loja= df_cidade_loja.sort_values(by='Valor de Venda',ascending= False)
df_cidade_loja
```

```python
df_estado_loja = df_caramelo.groupby(['Cidade Loja'])['Valor de Venda'].sum().reset_index()
df_estado_loja= df_estado_loja.sort_values(by='Valor de Venda',ascending= False)
df_estado_loja
```

```python
df_categoria_produto = df_caramelo.groupby(['Categoria Produto'])['Valor de Venda'].mean().reset_index()
df_categoria_produto= df_categoria_produto.sort_values(by='Valor de Venda',ascending= False)
df_categoria_produto
```

```python
df_eletronico = df_caramelo.loc[df_caramelo['Categoria Produto'] == 'Eletrônicos']
```

```python
df_categoria_eletronico = df_eletronico.groupby(['Marca'])['Valor de Venda'].sum().reset_index()
df_categoria_eletronico = df_categoria_eletronico.sort_values(by='Valor de Venda',ascending= False)
df_categoria_eletronico
```

```python
df_categoria_eletronico.plot(kind='bar', x= 'Marca',y='Valor de Venda')
plt.xlabel('Marcas Produto')
plt.ylabel('Valor de Venda')
plt.title('Valor de Venda por Marca')
plt.show()
```
